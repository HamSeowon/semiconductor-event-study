```python
import pandas as pd

price = pd.read_csv("../data/raw/multi_stock_price.csv", dtype = {"ticker":str})
bench = pd.read_csv("../data/raw/benchmarks.csv", dtype = {"ticker":str})

print(price.shape)
print(price['ticker'].unique())

print(bench.shape)
print(bench['ticker'].unique())
```

```python
print(price.info())
print(bench.info())
```

```python
price['Date'] = pd.to_datetime(price['Date'])
bench['Date'] = pd.to_datetime(bench['Date'])
```

```python
print(price.isna().sum())
print(bench.isna().sum())
```

```python
price = price.drop(columns = ['Change','Adj Close'])
```

```python
bench = bench.drop(columns = ['UpDown','Comp','Change','Amount','MarCap','Adj Close'])
```

```python
print(price.isna().sum())
print(bench.isna().sum())
```

```python
print(price.duplicated(subset=['Date', 'ticker']).sum())
print(bench.duplicated(subset=['Date', 'ticker']).sum())
```

```python
price = price.sort_values(['ticker', 'Date']).reset_index(drop=True)
bench = bench.sort_values(['ticker', 'Date']).reset_index(drop=True)

print(price.head())
print(bench.head())
```

```python
price['daily_return'] = price.groupby('ticker')['Close'].pct_change() * 100
bench['daily_return'] = bench.groupby('ticker')['Close'].pct_change() * 100
```

```python
print(price[price['ticker'] == '000660'][['Date', 'Close', 'daily_return']].head())
print(price[price['ticker'] == '005930'][['Date', 'Close', 'daily_return']].head())
print(price[price['ticker'] == 'MU'][['Date', 'Close', 'daily_return']].head())

```

```python
print(bench[bench['ticker'] == 'KOSPI'][['Date', 'Close', 'daily_return']].head())
print(bench[bench['ticker'] == 'NASDAQ'][['Date', 'Close', 'daily_return']].head())
```

```python
for ticker in price['ticker'].unique():
    subset = price[price['ticker'] == ticker]
    print(subset.nlargest(5, 'daily_return')[['Date', 'Close', 'daily_return']])
```

```python
for ticker in bench['ticker'].unique():
    subset = bench[bench['ticker'] == ticker]
    print(subset.nlargest(5, 'daily_return')[['Date', 'Close', 'daily_return']])
```

```python
price.columns = price.columns.str.lower()
bench.columns = bench.columns.str.lower()
```

```python
print(price.duplicated(subset=['date','ticker']).sum()) 
print(bench.duplicated(subset=['date','ticker']).sum()) 
```

```python
price.to_csv("../data/processed/multi_stock_price_clean.csv", index=False)
bench.to_csv("../data/processed/bench_clean.csv", index=False)
```
