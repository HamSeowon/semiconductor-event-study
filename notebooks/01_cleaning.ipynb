```python
import pandas as pd

price = pd.read_csv("../data/raw/sk_hynix_price.csv")
kospi = pd.read_csv("../data/raw/kospi.csv")

print(price.shape,kospi.shape)
price.head()
```

```python
price.info()
```

```python
kospi.info()
```

```python
price['Date'] = pd.to_datetime(price['Date'])
kospi['Date'] = pd.to_datetime(kospi['Date'])

print(price['Date'].dtype)
```

```python
print(price.isna().sum())
print(kospi.isna().sum())
```

```python
print(price['Date'].duplicated().sum())
print(kospi['Date'].duplicated().sum())
```

```python
print(price['Date'].is_monotonic_increasing)
print(kospi['Date'].is_monotonic_increasing)
```

```python
price['daily_return'] = price['Close'].pct_change() * 100 #how much percentage it has risen
price[['Date','Close','daily_return']].head(10) #why double bracket?
```

```python
price.nlargest(10, 'daily_return')[['Date','Close','daily_return']] #10 largest 
price.nsmallest(10, 'daily_return')[['Date','Close','daily_return']] #10 smallest
```

```python
kospi['daily_return'] = kospi['Close'].pct_change() * 100 #how much percentage it has risen
kospi[['Date','Close','daily_return']].head(10) #why double bracket?
```

```python
kospi.nlargest(10, 'daily_return')[['Date','Close','daily_return']] #10 largest 
kospi.nsmallest(10, 'daily_return')[['Date','Close','daily_return']] #10 smallest
```

```python
price.to_csv("../data/processed/sk_hynix_price_clean.csv", index = False)
kospi.to_csv("../data/processed/kospi_clean.csv", index = False)
```
